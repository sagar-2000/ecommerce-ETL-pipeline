```python
# Extract features for BI reporting
df_clean['month'] = df_clean['purchase_date'].dt.month
df_clean['day_of_week'] = df_clean['purchase_date'].dt.day_name()
```

```python
Q1 = df_clean['price'].quantile(0.25)
Q3 = df_clean['price'].quantile(0.75)
IQR = Q3 - Q1
outliers = df_clean[(df_clean['price'] < (Q1 - 1.5 * IQR)) | (df_clean['price'] > (Q3 + 1.5 * IQR))]
```

```python
df_ml = df_clean.drop(columns=['user_id', 'product_id', 'purchase_date', 'day_of_week', 'month' ])
```

```python
df_ml = pd.get_dummies(df_ml, columns=['category', 'payment_method'])
```

```python
df_ml
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Split 80% for building clusters and 20% for validation
train_df, val_df = train_test_split(df_ml, test_size=0.2, random_state=42)

# Scale based ONLY on the training data
scaler = StandardScaler()
train_scaled = scaler.fit_transform(train_df)

# Apply that same scaling to the validation set (no fitting here!)
val_scaled = scaler.transform(val_df)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(train_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 4))
plt.plot(K_range, inertia, 'bx-')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia (Sum of squared distances)')
plt.title('The Elbow Method showing the optimal k')
```

```python
# 1. Fit the final model with k=7
kmeans_final = KMeans(n_clusters=7, random_state=42, n_init=10, algorithm="lloyd")
train_df['Cluster'] = kmeans_final.fit_predict(train_scaled)

# 2. Group by Cluster to see the "Average" customer in each group
# We use the unscaled data for this so the numbers make sense (Euros, Percentages)
cluster_profile = train_df.groupby('Cluster')[['price', 'discount_pct', 'final_price']].mean()

print(cluster_profile)
```
